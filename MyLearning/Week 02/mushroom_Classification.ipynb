মাশরুম ক্লাসিফিকেশন

```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("default")
sns.set(font_scale=1.1)

# ফাইল লোড
df = pd.read_csv("mushrooms.csv")

print("ডেটার সাইজ →", df.shape)
print("\nপ্রথম ৫টা রো:")
display(df.head())

print("কলাম ও টাইপের তথ্য:")
df.info()

print("\nমিসিং ভ্যালু গণনা:")
print(df.isnull().sum())

print("\nক্যাটাগরিক্যাল সারাংশ (সব কলাম ক্যাটাগরিক্যাল):")
display(df.describe(include='all'))
```

## ধাপ ৪: টার্গেট ভ্যারিয়েবল বোঝা (class) 🎯

```python
# ১. সংখ্যায় count
print("ক্লাসের বিতরণ:")
print(df['class'].value_counts())
# e: 4208, p: 3916

# ২. Percentage এ
print("\nপার্সেন্টেজে:")
print(df['class'].value_counts(normalize=True) * 100)
# e: 51.8%, p: 48.2%

# ৩. Visualize
plt.figure(figsize=(6,4))                    # নতুন plot
sns.countplot( x='class', data=df, palette='viridis')  # বার চার্ট
plt.title("Edible vs Poisonous Mushrooms")   # Title
plt.xlabel("Class (e = Edible, p = Poisonous)")  # X-label
plt.ylabel("Count")                          # Y-label
plt.show()                                   # Display
```

## ধাপ ৫: ক্যাটাগরিক্যাল কলামগুলোর ফ্রিকোয়েন্সি দেখা 📊

```python
cat_cols = df.columns[1:]  # class বাদ দিয়ে সব ফিচার

for col in cat_cols:
    print(f"\n{col} এর বিতরণ:")
    print(df[col].value_counts())
    print(df[col].value_counts(normalize=True) * 100)

    # ভিজুয়াল (যদি অনেক ক্যাটাগরি না থাকে)
    if df[col].nunique() <= 10:
        plt.figure(figsize=(8,4))
        sns.countplot(x=col, data=df, order=df[col].value_counts().index)
        plt.title(f"{col} Distribution")
        plt.xticks(rotation=45)
        plt.show()
```

## ধাপ ৬: টার্গেটের সাথে ক্যাটাগরিক্যাল ফিচারের সম্পর্ক দেখা 🔗

```python
important_cat = ['cap-shape', 'cap-surface', 'cap-color', 'bruises', 'odor', 'gill-attachment', 'gill-spacing', 'gill-size', 'gill-color', 'stalk-shape', 'stalk-root']

for col in important_cat:
    plt.figure(figsize=(8,5))
    sns.countplot(x=col, hue='class', data=df)
    plt.title(f"{col} vs Class (e/p)")
    plt.xticks(rotation=45)
    plt.show()
```

## ধাপ ৯: Preprocessing (ডেটা পরিষ্কার ও প্রস্তুতি) 🔧

```python
# ========================================
# প্রথমে column names দেখো
# ========================================

print("Available columns:")
print(df.columns.tolist())
print("\nColumn with 'stalk' or 'root':")
print([col for col in df.columns if 'stalk' in col.lower() or 'root' in col.lower()])


# ========================================
# Part 1: Missing Values Handle (FIXED)
# ========================================

# Option 1: যদি column name আলাদা হয় (underscore দিয়ে)
if 'stalk_root' in df.columns:
    df['stalk_root'] = df['stalk_root'].replace('?', np.nan)
    df['stalk_root'].fillna(df['stalk_root'].mode()[0], inplace=True)

# Option 2: যদি hyphen দিয়ে হয় কিন্তু load হয়নি
elif 'stalk-root' in df.columns:
    df['stalk-root'] = df['stalk-root'].replace('?', np.nan)
    df['stalk-root'].fillna(df['stalk-root'].mode()[0], inplace=True)

# Option 3: সব columns এ '?' খুঁজে replace করো (safest)
else:
    print("'stalk-root' column not found. Handling all '?' in dataset...")
    for col in df.columns:
        if df[col].dtype == 'object':  # শুধু text columns
            if '?' in df[col].values:
                df[col] = df[col].replace('?', np.nan)
                df[col].fillna(df[col].mode()[0], inplace=True)
                print(f"Fixed missing values in: {col}")


# ========================================
# Part 2: Categorical Encoding
# ========================================

# Method A: One-Hot Encoding
cat_cols = [col for col in df.columns if col != 'class']  # class বাদে সব
df = pd.get_dummies(df, columns=cat_cols, drop_first=True)

# টার্গেট এনকোড
df['class'] = df['class'].map({'e':0, 'p':1})

print("\n✅ Preprocessing complete!")
print(f"Final shape: {df.shape}")
print(f"Missing values: {df.isnull().sum().sum()}")
```

## ধাপ ১০: Feature Engineering (নতুন ফিচার তৈরি) 🎨

```python
# ========================================
# শুরু থেকে সঠিক order এ
# ========================================

import pandas as pd
import numpy as np

# Step 1: Load data
df = pd.read_csv("mushrooms.csv")

# Step 2: Handle missing
df = df.replace('?', np.nan)
for col in df.columns:
    if df[col].isnull().sum() > 0:
        df[col].fillna(df[col].mode()[0], inplace=True)


# Step 3: Feature Engineering (BEFORE encoding)
print("Creating new features...")

# 1. Bad odor flag
df['bad_odor'] = df['odor'].isin(['f', 'c', 'y', 's', 'p', 'm']).astype(int)

# 2. No smell (safest)
df['no_smell'] = (df['odor'] == 'n').astype(int)

# 3. Almond/anise smell (good signs)
df['good_smell'] = df['odor'].isin(['a', 'l']).astype(int)

# 4. Bruises present
df['has_bruises'] = (df['bruises'] == 't').astype(int)

# 5. White spore print
df['white_spores'] = (df['spore-print-color'] == 'w').astype(int)

# 6. Broad gills
df['broad_gills'] = (df['gill-size'] == 'b').astype(int)

# 7. Gill spacing
df['close_gills'] = (df['gill-spacing'] == 'c').astype(int)

# 8. Population type (abundant = common, safer?)
df['abundant'] = (df['population'] == 'v').astype(int)

# 9. Habitat (woods vs urban)
df['in_woods'] = (df['habitat'] == 'd').astype(int)

# 10. Ring number (multiple rings)
df['multiple_rings'] = (df['ring-number'] == 't').astype(int)

# 11. Veil color (white veil)
df['white_veil'] = (df['veil-color'] == 'w').astype(int)

# 12. Interaction features
df['odor_gill_interaction'] = df['bad_odor'] * df['broad_gills']
df['smell_bruise_combo'] = df['no_smell'] * df['has_bruises']


print("✅ Created features:")
new_features = ['bad_odor', 'no_smell', 'good_smell', 'has_bruises',
                'white_spores', 'broad_gills', 'close_gills', 'abundant',
                'in_woods', 'multiple_rings', 'white_veil',
                'odor_gill_interaction', 'smell_bruise_combo']

print(df[new_features].head())
print(f"\nNew feature counts:\n{df[new_features].sum()}")


# Step 4: Now encode
cat_cols = [col for col in df.columns if col not in ['class'] + new_features]
df = pd.get_dummies(df, columns=cat_cols, drop_first=True)

# Target encode
df['class'] = df['class'].map({'e':0, 'p':1})

print(f"\n✅ Final shape: {df.shape}")
print(f"✅ Total features: {df.shape[1] - 1}")  # -1 for class
```
