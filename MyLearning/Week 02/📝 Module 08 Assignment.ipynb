# 🔬 স্তম্ভ ৩ — Preprocessing (Data পরিষ্কার করা)

## Core দর্শন:
> **"Garbage in, Garbage out।"**
> নোংরা data দিলে model ও নোংরা উত্তর দেবে।

Preprocessing এর তিনটা কাজ —

### কাজ ১: Missing Values — ফাঁকা জায়গা পূরণ করা

```
কল্পনা করো পরীক্ষার খাতায় কিছু প্রশ্নের উত্তর ফাঁকা।
Examiner কী করবে? দেখবে বাকিরা কী লিখেছে
এবং সেই অনুযায়ী একটা reasonable উত্তর বসাবে।

Numeric (Age)     → Median বসাও (গড় বয়স)
Categorical (Embarked) → Mode বসাও (সবচেয়ে common port)
৭৭% ফাঁকা (Cabin) → বাদ দিয়ে দাও (এটা রেখে লাভ নেই)
```

### কাজ ২: Encoding — ভাষা বদলানো

```
Machine শুধু সংখ্যা বোঝে।
"male/female" বোঝে না।

এটা ঠিক যেমন —
তুমি বাংলায় কথা বলছো, machine বোঝে শুধু ইংরেজি।
তাই translate করতে হয়।

"male"   → 0
"female" → 1

"S" → 0, "C" → 1, "Q" → 2  (OneHotEncoding দিয়ে আলাদা column)
```

### কাজ ৩: Scaling — সবাইকে এক মাঠে আনা

```
কল্পনা করো দুটো খেলোয়াড়ের তুলনা করছো —
একজনের উচ্চতা ফুটে মাপা (6 ফুট)
আরেকজনের ওজন কেজিতে মাপা (70 কেজি)

এদের সরাসরি তুলনা করা যায় না!
Scaling করলে সবাই একই unit এ আসে।

Age  → 0 to 80   ⟐ Scaling ⟶  -2 to +2
Fare → 0 to 512  ⟐ Scaling ⟶  -2 to +2
```

# ⚙️ Setup Cell

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder, StandardScaler

sns.set(style="whitegrid")
plt.rcParams["figure.figsize"] = (8, 5)

print("Libraries imported.")
````

````python
df = pd.read_csv("Titanic-Dataset .csv")

print("Shape of dataset:", df.shape)
display(df.head())

print("\nInfo:")
print(df.info())

print("\nDescribe (numeric columns):")
display(df.describe())
````

# Part A-1 — Exploration + Drop

# Part A-4 — Categorical vs Target

````python
# given — শুধু run করো
cat_target_cols = ["Sex", "Pclass", "Embarked"]

for c in cat_target_cols:
    ct = pd.crosstab(df[c], df["Survived"], normalize="index")
    print(f"\nSurvival proportion by {c}:")
    display(ct)
    ct.plot(kind="bar", stacked=True)
    plt.title(f"Survival proportion by {c}")
    plt.ylabel("Proportion")
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.show()
````

**Markdown Answer:**
```
- Insight 1:
  Female passengers had a survival rate of ~74% while
  male passengers had only ~19%, clearly confirming
  the "women first" evacuation policy.

- Insight 2:
  1st class passengers survived at ~63% while 3rd
  class survived at only ~24%, showing that social
  class significantly affected survival chances.

# Part B-5 — Missing Values

````python
print("Missing values before:")
print(df.isna().sum())

# Drop Cabin
df = df.drop('Cabin', axis=1, errors='ignore')

# Fill Age with Median
age_median = df['Age'].median()
df['Age'] = df['Age'].fillna(age_median)

# Fill Embarked with Mode
embarked_mode = df['Embarked'].mode()[0]
df['Embarked'] = df['Embarked'].fillna(embarked_mode)

print("\nMissing values after handling:")
print(df.isna().sum())
````

**Markdown Answer:**
```
- Why you dropped or kept Cabin:
  Cabin was dropped because over 77% of its values
  are missing. Imputing such a large proportion would
  introduce more noise than signal into the model.

- Why you used median/mean for Age:
  Median was used for Age because it is robust to
  outliers. Age has some very old passengers that
  could skew the mean, making median a safer choice.

- Why you used mode for Embarked:
  Embarked is a categorical variable so mean and
  median are not applicable. Filling with the most
  frequent value (mode = 'S') is the standard
  approach for categorical missing data.

# Part B-6 — Encoding

````python
df_enc = df.copy()

cat_features = ["Sex", "Embarked", "Pclass"]
numeric_features = ["Age", "Fare", "SibSp", "Parch"]
target_col = "Survived"

# OneHotEncoder — TODO
ohe = OneHotEncoder(sparse_output=False, drop='first')
encoded_array = ohe.fit_transform(df_enc[cat_features])
encoded_df = pd.DataFrame(
    encoded_array,
    columns=ohe.get_feature_names_out(cat_features)
)

# Final DataFrame — given
df_model = pd.concat([
    df_enc[[target_col] + numeric_features].reset_index(drop=True),
    encoded_df.reset_index(drop=True)
], axis=1)

print("Shape after encoding:", df_model.shape)
display(df_model.head())
````

# Part B-7 — Scaling

````python
score_cols = ["Age", "Fare", "SibSp", "Parch"]

# Boxplot before — given
plt.figure(figsize=(6, 4))
sns.boxplot(data=df_model[score_cols])
plt.title("Numeric features before scaling")
plt.tight_layout()
plt.show()

# StandardScaler — TODO
scaler = StandardScaler()
df_scaled = df_model.copy()
df_scaled[score_cols] = scaler.fit_transform(df_model[score_cols])

# Boxplot after — given
plt.figure(figsize=(6, 4))
sns.boxplot(data=df_scaled[score_cols])
plt.title("Numeric features after StandardScaler")
plt.tight_layout()
plt.show()
````

**Markdown Answer:**
```
- Sentence 1:
  StandardScaler transforms each numeric feature to
  have mean = 0 and standard deviation = 1, bringing
  all features to the same scale.

- Sentence 2:
  Without scaling, distance-based models like KNN or
  SVM would incorrectly give more importance to
  features with larger ranges such as Fare compared
  to smaller ones like SibSp.
