```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier, plot_tree
import matplotlib.pyplot as plt

# Dataset তৈরি
data = {
    'Weather': ['Sunny','Rainy','Sunny','Sunny','Rainy','Rainy','Sunny','Rainy'],
    'Windy':   [0, 1, 0, 1, 0, 1, 0, 1],
    'Play':    [1, 0, 1, 1, 0, 0, 1, 0]
}
df = pd.DataFrame(data)

# Weather encode করা
df['Weather_num'] = df['Weather'].map({'Sunny': 1, 'Rainy': 0})
X = df[['Weather_num', 'Windy']]
y = df['Play']

# TODO: DecisionTreeClassifier তৈরি ও fit
model = DecisionTreeClassifier(max_depth=2, random_state=42)
model.fit(X, y)

# Tree visualize
plt.figure(figsize=(10, 5))
plot_tree(
    model,
    feature_names=X.columns,
    class_names=['No Play', 'Play'],
    filled=True,
    rounded=True,
    fontsize=12
)
plt.title("Decision Tree — Weather vs Play")
plt.show()
```
