চল আমরা এটা একদম শূন্য থেকে, “কেন” এবং “কিভাবে”—এই দুইটা দৃষ্টিকোণ থেকে শিখি। তুমি শুধু ফর্মুলা মুখস্থ না করে, এর পেছনের বিজ্ঞানটা বুঝবে।

---

## 🧠 ১. Entropy আসলে কী?

Entropy মানে হলো **অনিশ্চয়তার পরিমাণ (uncertainty)**।

* যদি সবকিছু নিশ্চিত হয় → entropy = 0
* যদি কিছুই নিশ্চিত না হয় → entropy বেশি

👉 উদাহরণ:

* যদি তুমি জানো ১০০% যে ফলাফল “A” → কোনো uncertainty নেই → entropy = 0
* যদি ৫০%-৫০% chance → তুমি confused → entropy maximum

---

## 🔬 ২. কেন log ব্যবহার করা হয়?

Entropy-র ভিতরে `log₂` আছে—এটা কাকতালীয় না, এর পেছনে শক্ত বৈজ্ঞানিক কারণ আছে।

### 👉 কারণ:

1. **Information measure করতে log লাগে**

   * Rare event → বেশি information
   * Common event → কম information

2. log₂ মানে:
   👉 “কতগুলো yes/no (binary) প্রশ্ন লাগবে ফলাফল বের করতে?”

---

## 📐 ৩. মূল ফর্মুলা (Binary Case)

H = -p_1 \log_2(p_1) - p_2 \log_2(p_2)

---

## 🧩 ৪. ফর্মুলাটাকে ভেঙে বুঝি

### প্রতিটা অংশের মানে:

#### 🔹 p₁ log₂(p₁)

* p₁ = probability
* log₂(p₁) = information content

👉 Rare হলে (p ছোট):

* log₂(p) → বড় negative
* তাই information → বেশি

👉 Common হলে (p বড়):

* log₂(p) → কম negative
* information → কম

---

### 🔹 কেন minus (-) দেওয়া হয়?

কারণ:

* log₂(p) সবসময় negative (0 < p < 1 হলে)
* entropy সবসময় positive হওয়া উচিত

👉 তাই minus দিয়ে positive বানানো হয়

---

## 🧪 ৫. Scientific Experiment দিয়ে বুঝি

### Case 1: সম্পূর্ণ নিশ্চিত

p₁ = 1, p₂ = 0

Entropy =
= -1×log₂(1) - 0×log₂(0)
= -1×0 - 0
= **0**

👉 Interpretation: কোনো uncertainty নেই

---

### Case 2: সম্পূর্ণ অনিশ্চিত

p₁ = 0.5, p₂ = 0.5

Entropy =
= -0.5×log₂(0.5) - 0.5×log₂(0.5)
= -0.5×(-1) - 0.5×(-1)
= 0.5 + 0.5
= **1 (maximum)**

👉 Interpretation: maximum confusion

---

### Case 3: মাঝামাঝি

p₁ = 0.8, p₂ = 0.2

Entropy ≈ 0.72

👉 Interpretation:

* কিছুটা নিশ্চিত
* কিন্তু পুরোপুরি না

## 🧠 ৭. Machine Learning-এ কেন লাগে?

Decision Tree-তে entropy ব্যবহার হয়:

👉 Goal:
**ডাটা split করে uncertainty কমানো**

### Step:

1. Parent node → high entropy
2. Split করার পর → lower entropy
3. Best split = maximum entropy reduction

👉 এটাকেই বলে:
**Information Gain**

* যদি random YES/NO → অনেক প্রশ্ন লাগবে

👉 প্রশ্নের সংখ্যা ≈ entropy

---

# 🔷 Step 1: Formula টা কী বলছে?

H = -(p\log_2 p + (1-p)\log_2(1-p))

এখানে:

* (p) = Yes হওয়ার chance
* (1-p) = No হওয়ার chance

👉 এই formula আসলে **দুইটা জিনিস যোগ করছে**:

* Yes থেকে কত uncertainty
* No থেকে কত uncertainty

---

# 🔷 Step 2: Log কেন আছে (easy view)

তুমি শুধু এইটা মনে রাখো:

| Probability | log₂(p) |
| ----------- | ------- |
| 1           | 0       |
| 0.5         | -1      |
| 0.25        | -2      |

👉 probability যত ছোট → log তত negative
👉 মানে: **rare event = বেশি surprise**

---

# 🔷 Step 3: পুরো formula কী করছে?

ধরো (p = 0.5)

[
H = -(0.5 \cdot (-1) + 0.5 \cdot (-1))
]

Step:

* 0.5 × (-1) = -0.5
* দুইটা add = -1
* সামনে minus → **1**

👉 তাই entropy = 1 (max confusion)

---

# 🔷 Step 4: Pure case (সব same)

ধরো (p = 1)

[
H = -(1\cdot 0 + 0\cdot \log_2 0)
]

👉 এখানে:

* log₂(1) = 0
* (0 \cdot \log(0)) → treat as 0

👉 result = 0

✔️ কোনো confusion নেই

---

# 🔷 Step 5: মাঝামাঝি case

ধরো (p = 0.75)

[
H = -(0.75 \cdot (-0.415) + 0.25 \cdot (-2))
]

Step:

* 0.75 × (-0.415) ≈ -0.311
* 0.25 × (-2) = -0.5
* total = -0.811
* minus দিলে = **0.811**

👉 medium confusion

---

# 🔷 Step 6: এখন split বুঝি (most important)

### Rule:

[
\text{After split} = \sum (\text{subset size / total}) \times H(\text{subset})
]

👉 মানে: weighted average

---

# 🔷 Weather example

* Sunny → H = 0
* Rainy → H = 0

[
H = \frac{2}{4}\cdot 0 + \frac{2}{4}\cdot 0 = 0
]

👉 পুরো confusion শেষ ✅

---

# 🔷 Windy example

* Windy=No → H = 1
* Windy=Yes → H = 1

[
H = \frac{2}{4}\cdot 1 + \frac{2}{4}\cdot 1 = 1
]

👉 কোনো improvement নেই ❌

---

# 🔷 Final intuition (math + simple)

* log → surprise measure
* p × log(p) → weighted surprise
* sum → total uncertainty
* minus → positive value বানায়

---

# 🔚 এক লাইনে

👉 **Entropy = weighted average surprise**

চল এটাকে একদম শিক্ষক-স্টাইলে, শুরু থেকে, “কী”, “কেন”, “কিভাবে” — এই তিন স্তরে বুঝি।
তুমি যে শব্দটা বারবার দেখছো, **Gini** বা **Gini impurity**, সেটা আসলে decision tree-তে একটি node কতটা “মিশ্র” বা “অশুদ্ধ” তা মাপে।

## 1) Gini আসলে কী?

ধরো একটি node-এ শুধু দুই class আছে: Yes এবং No.

* যদি node-এর সব sample Yes হয়, তাহলে node **pure**
* যদি node-এ Yes আর No মিশে থাকে, তাহলে node **impure**

**Gini impurity** সেই মিশ্রণ কতটা, তা মাপে।

সবচেয়ে সহজ intuition:

> একটি node থেকে random দুইটা item নিলে, তারা different class হওয়ার সম্ভাবনা কত?

যত বেশি different হওয়ার সম্ভাবনা, node তত বেশি mixed, তত বেশি impurity.

---

## 2) কেন এটা “অপবিত্রতা” মাপে?

কারণ decision tree-এর লক্ষ্য হলো এমন split করা, যাতে child node-গুলোতে এক class-এর sample বেশি থাকে।

ধরো root node-এ 50% Yes, 50% No।
এখানে model নিশ্চিতভাবে কিছু বলতে পারছে না।
এটা একটি **high impurity** node।

আর যদি node-এ 100% Yes থাকে, তাহলে model পুরোপুরি নিশ্চিত।
এটা **zero impurity**।

সুতরাং impurity মানে হলো:

* uncertainty
* mixture
* confusion
* class overlap

---

## 3) Gini formula কেন এমন?

Binary case-এ:

[
Gini = 1 - (p_1^2 + p_2^2)
]

এখানে:

* (p_1) = class 1-এর probability
* (p_2) = class 2-এর probability
* (p_1 + p_2 = 1)

### এই সূত্রের intuition

ধরো তুমি node থেকে **দুইটা item** randomly নিলে।

* প্রথমটা class 1 হওয়ার সম্ভাবনা = (p_1)
* দ্বিতীয়টাও class 1 হওয়ার সম্ভাবনা = (p_1)

তাহলে দুটোই class 1 হওয়ার সম্ভাবনা = (p_1^2)

একইভাবে দুটোই class 2 হওয়ার সম্ভাবনা = (p_2^2)

সুতরাং, **দুটো item একই class-এর হওয়ার probability**:

[
p_1^2 + p_2^2
]

এটা node-এর **purity** বোঝায়।

আর impurity চাইলে তার complement:

[
1 - (p_1^2 + p_2^2)
]

এটাই Gini।

অর্থাৎ:

* same-class pair probability বেশি → node cleaner → Gini কম
* different/mixed nature বেশি → node dirtier → Gini বেশি

---

## 4) Binary case-এ একে আরেকভাবে লেখা যায়

যেহেতু (p_2 = 1 - p_1),

[
Gini = 1 - p_1^2 - (1-p_1)^2
]

এটা expand করলে:

[
Gini = 2p_1(1-p_1)
]

এখান থেকে একটা সুন্দর insight আসে:

* (p_1 = 0) বা (1) হলে Gini = 0
* (p_1 = 0.5) হলে Gini = 0.5

অর্থাৎ binary classification-এ Gini সর্বোচ্চ হয় **৫০-৫০ mixed** অবস্থায়।

---

## 5) উদাহরণ দিয়ে বোঝা যাক

### Case A: Pure node

ধরো সব Yes।

* (p(Yes)=1)
* (p(No)=0)

[
Gini = 1 - (1^2 + 0^2) = 0
]

এখানে uncertainty নেই, তাই impurity শূন্য।

---

### Case B: Half-half node

ধরো 50% Yes, 50% No।

* (p(Yes)=0.5)
* (p(No)=0.5)

[
Gini = 1 - (0.5^2 + 0.5^2)
= 1 - (0.25 + 0.25)
= 0.5
]

এটাই binary case-এ maximum impurity।

---

### Case C: Mostly one class

ধরো 90% Yes, 10% No।

[
Gini = 1 - (0.9^2 + 0.1^2)
= 1 - (0.81 + 0.01)
= 0.18
]

এখানে node অনেক বেশি pure, তাই Gini কম।

---

## 6) এটা scientificভাবে কীভাবে ভাববে?

Scientific method-এ আমরা সাধারণত এইভাবে চিন্তা করি:

### Observation

ডাটা mixed থাকলে prediction কঠিন, pure থাকলে সহজ।

### Hypothesis

যে split child node-গুলোর impurity কমাবে, সেটাই ভালো split হবে।

### Experiment

প্রতিটি সম্ভাব্য split-এর পর Gini calculate করি।

### Decision

যে split-এর **weighted Gini** সবচেয়ে কম, সেটি বেছে নিই।

---

## 7) Split-এর Gini কীভাবে বের হয়?

একটা split-এর পরে left child আর right child তৈরি হলো।

তাহলে total split impurity:

[
Gini_{split} = \frac{n_L}{n}Gini_L + \frac{n_R}{n}Gini_R
]

এখানে:

* (n_L) = left child-এর sample সংখ্যা
* (n_R) = right child-এর sample সংখ্যা
* (n) = total sample

### Rule:

**যে split-এর weighted Gini কম, সেটি better.**

---

## 8) Entropy আর Gini—দুটোই কেন আছে?

দুটোই impurity মাপে, কিন্তু slightly different lens দিয়ে।

### Entropy

[
H = -\sum p_i \log_2(p_i)
]

### Gini

[
Gini = 1 - \sum p_i^2
]

দুইটার common idea:

* pure node → 0
* mixed node → বড় value

---

## 9) পার্থক্য কোথায়?

### Entropy

* log ব্যবহার করে
* uncertainty-কে information-theoreticভাবে মাপে
* একটু বেশি mathematically “sensitive”

### Gini

* square ব্যবহার করে
* simpler
* computationally cheaper
* practical decision trees-এ খুব জনপ্রিয়

---

## 10) Binary case-এ তুলনা

| p(Yes) | Entropy |  Gini |
| ------ | ------: | ----: |
| 0.5    |   1.000 | 0.500 |
| 0.9    |   0.469 | 0.180 |
| 1.0    |   0.000 | 0.000 |

### Meaning

* দুটোই বলছে 50-50 node সবচেয়ে mixed
* Entropy সাধারণত impurity-র extreme behavior-এ একটু বেশি sensitive
* Gini একটু smoother, simpler

---

## 11) Gini-র গভীর intuition: কেন square?

Square করার ফলে বড় probability-এর contribution বেশি weight পায়।

ধরো:

* (0.9^2 = 0.81)
* (0.1^2 = 0.01)

Major class dominance দ্রুত ধরা পড়ে।
তাই Gini node-এর class concentration খুব naturally capture করে।

---

## 12) আরেকটা দারুণ connection: variance

Binary case-এ যদি class label 0/1 হিসেবে ভাবো, তাহলে Bernoulli variance:

[
Var(Y) = p(1-p)
]

আর Gini:

[
Gini = 2p(1-p)
]

অর্থাৎ Gini variance-এর খুব কাছাকাছি এক ধরনের measure।
এই কারণে এটাকে “spread/mixture” বোঝার একটি খুব intuitive metric বলা যায়।

---

## 13) Decision tree কেন এটা ব্যবহার করে?

Decision tree-এর মূল লক্ষ্য:

> এমন feature-based split খোঁজা, যার ফলে child nodes-এ class গুলো যতটা সম্ভব clean হয়।

মানে:

* root mixed
* split করে clean child বানানো
* repeated split
* শেষে leaf node pure বা near-pure

Gini এই purity improvement মাপে।

---

## 14) এক লাইনে মনে রাখার মতো intuition

**Gini impurity = “node-এর ভেতরে class mix কতটা?”**

* সব sample এক class → 0
* 50-50 mix → বেশি
* মাঝামাঝি mix → মাঝামাঝি value

---

## 15) তোমার দেয়া “Entropy vs Gini” তুলনাটা কীভাবে পড়বে?

তুমি লিখেছো:

* p(Yes)=0.5 → Entropy 1.0, Gini 0.5
* p(Yes)=0.9 → Entropy 0.469, Gini 0.18

এটার অর্থ:

### 50-50 node

দুই metric-ই বলছে: “এটা খুব mixed”

### 90-10 node

দুই metric-ই বলছে: “এটা বেশ pure”

তফাৎ শুধু মাপার স্কেল আর sensitivity-তে।

---

## 16) শেষ mental model

ভাবো node-এর মধ্যে sample গুলো students:

* সব একই uniform পরা → pure
* random mixed uniforms → impure

Gini প্রশ্ন করছে:

> “এই node থেকে random দুইজন নিলে, তারা একই ধরনের কি?”

যদি উত্তর “হ্যাঁ, বেশিরভাগ সময়” হয়, impurity কম।
যদি উত্তর “না, প্রায়ই আলাদা” হয়, impurity বেশি।

https://chatgpt.com/s/t_69ec187670f88191b672ff152bb2aba9



## 1) Gini আসলে কী?

ধরো একটি node-এ শুধু দুই class আছে: Yes এবং No.

* যদি node-এর সব sample Yes হয়, তাহলে node **pure**
* যদি node-এ Yes আর No মিশে থাকে, তাহলে node **impure**

**Gini impurity** সেই মিশ্রণ কতটা, তা মাপে।

সবচেয়ে সহজ intuition:

> একটি node থেকে random দুইটা item নিলে, তারা different class হওয়ার সম্ভাবনা কত?

যত বেশি different হওয়ার সম্ভাবনা, node তত বেশি mixed, তত বেশি impurity.

---

## 2) কেন এটা “অপবিত্রতা” মাপে?

কারণ decision tree-এর লক্ষ্য হলো এমন split করা, যাতে child node-গুলোতে এক class-এর sample বেশি থাকে।

ধরো root node-এ 50% Yes, 50% No।
এখানে model নিশ্চিতভাবে কিছু বলতে পারছে না।
এটা একটি **high impurity** node।

আর যদি node-এ 100% Yes থাকে, তাহলে model পুরোপুরি নিশ্চিত।
এটা **zero impurity**।

সুতরাং impurity মানে হলো:

* uncertainty
* mixture
* confusion
* class overlap

---

## 3) Gini formula কেন এমন?

Binary case-এ:

[
$$Gini = 1 - (p_1^2 + p_2^2)$$
]

এখানে:

* (p_1) = class 1-এর probability
* (p_2) = class 2-এর probability
* (p_1 + p_2 = 1)

### এই সূত্রের intuition

ধরো তুমি node থেকে **দুইটা item** randomly নিলে।

* প্রথমটা class 1 হওয়ার সম্ভাবনা = (p_1)
* দ্বিতীয়টাও class 1 হওয়ার সম্ভাবনা = (p_1)

তাহলে দুটোই class 1 হওয়ার সম্ভাবনা = (p_1^2)

একইভাবে দুটোই class 2 হওয়ার সম্ভাবনা = (p_2^2)

সুতরাং, **দুটো item একই class-এর হওয়ার probability**:

[
p_1^2 + p_2^2
]

এটা node-এর **purity** বোঝায়।

আর impurity চাইলে তার complement:

[
1 - (p_1^2 + p_2^2)
]

এটাই Gini।

অর্থাৎ:

* same-class pair probability বেশি → node cleaner → Gini কম
* different/mixed nature বেশি → node dirtier → Gini বেশি

---

## 4) Binary case-এ একে আরেকভাবে লেখা যায়

যেহেতু (p_2 = 1 - p_1),

[
Gini = 1 - p_1^2 - (1-p_1)^2
]

এটা expand করলে:

[
Gini = 2p_1(1-p_1)
]

এখান থেকে একটা সুন্দর insight আসে:

* (p_1 = 0) বা (1) হলে Gini = 0
* (p_1 = 0.5) হলে Gini = 0.5

অর্থাৎ binary classification-এ Gini সর্বোচ্চ হয় **৫০-৫০ mixed** অবস্থায়।

---

## 5) উদাহরণ দিয়ে বোঝা যাক

### Case A: Pure node

ধরো সব Yes।

* (p(Yes)=1)
* (p(No)=0)

[
Gini = 1 - (1^2 + 0^2) = 0
]

এখানে uncertainty নেই, তাই impurity শূন্য।

---

### Case B: Half-half node

ধরো 50% Yes, 50% No।

* (p(Yes)=0.5)
* (p(No)=0.5)

[
Gini = 1 - (0.5^2 + 0.5^2)
= 1 - (0.25 + 0.25)
= 0.5
]

এটাই binary case-এ maximum impurity।

---

### Case C: Mostly one class

ধরো 90% Yes, 10% No।

[
Gini = 1 - (0.9^2 + 0.1^2)
= 1 - (0.81 + 0.01)
= 0.18
]

এখানে node অনেক বেশি pure, তাই Gini কম।

---

## 6) এটা scientificভাবে কীভাবে ভাববে?

Scientific method-এ আমরা সাধারণত এইভাবে চিন্তা করি:

### Observation

ডাটা mixed থাকলে prediction কঠিন, pure থাকলে সহজ।

### Hypothesis

যে split child node-গুলোর impurity কমাবে, সেটাই ভালো split হবে।

### Experiment

প্রতিটি সম্ভাব্য split-এর পর Gini calculate করি।

### Decision

যে split-এর **weighted Gini** সবচেয়ে কম, সেটি বেছে নিই।

---

## 7) Split-এর Gini কীভাবে বের হয়?

একটা split-এর পরে left child আর right child তৈরি হলো।

তাহলে total split impurity:

[
Gini_{split} = \frac{n_L}{n}Gini_L + \frac{n_R}{n}Gini_R
]

এখানে:

* (n_L) = left child-এর sample সংখ্যা
* (n_R) = right child-এর sample সংখ্যা
* (n) = total sample

### Rule:

**যে split-এর weighted Gini কম, সেটি better.**

---

## 8) Entropy আর Gini—দুটোই কেন আছে?

দুটোই impurity মাপে, কিন্তু slightly different lens দিয়ে।

### Entropy

[
H = -\sum p_i \log_2(p_i)
]

### Gini

[
Gini = 1 - \sum p_i^2
]

দুইটার common idea:

* pure node → 0
* mixed node → বড় value

---

## 9) পার্থক্য কোথায়?

### Entropy

* log ব্যবহার করে
* uncertainty-কে information-theoreticভাবে মাপে
* একটু বেশি mathematically “sensitive”

### Gini

* square ব্যবহার করে
* simpler
* computationally cheaper
* practical decision trees-এ খুব জনপ্রিয়

| ------ | ------: | ----: |

| 0.9    |   0.469 | 0.180 |
| 1.0    |   0.000 | 0.000 |

* Gini একটু smoother, simpler

## 11) Gini-র গভীর intuition: কেন square?

Square করার ফলে বড় probability-এর contribution বেশি weight পায়।

ধরো:

* (0.9^2 = 0.81)
* (0.1^2 = 0.01)

Major class dominance দ্রুত ধরা পড়ে।
তাই Gini node-এর class concentration খুব naturally capture করে।

## 12) আরেকটা দারুণ connection: variance

Binary case-এ যদি class label 0/1 হিসেবে ভাবো, তাহলে Bernoulli variance:

[
Var(Y) = p(1-p)
]

আর Gini:

[
Gini = 2p(1-p)
]

অর্থাৎ Gini variance-এর খুব কাছাকাছি এক ধরনের measure।
এই কারণে এটাকে “spread/mixture” বোঝার একটি খুব intuitive metric বলা যায়।

## 13) Decision tree কেন এটা ব্যবহার করে?

Decision tree-এর মূল লক্ষ্য:

> এমন feature-based split খোঁজা, যার ফলে child nodes-এ class গুলো যতটা সম্ভব clean হয়।

মানে:

* root mixed
* split করে clean child বানানো
* repeated split
* শেষে leaf node pure বা near-pure

Gini এই purity improvement মাপে।

# 🔬 3) Information Gain কী?

এখন মূল scientific idea:

## ❗ Hypothesis

যদি আমরা data split করি, তাহলে chaos কমবে।

## ❗ Test

Split করার পরে entropy কত কমল?

## ❗ Measure

IG = H(\text{parent}) - \sum_{i} \frac{n_i}{n} H(\text{child}_i)

---

## 🧠 এই formula-এর অর্থ (খুব গুরুত্বপূর্ণ)

### Left side:

👉 IG = “কতটা uncertainty কমল”

---

### Right side breakdown:

#### 1. H(parent)

👉 split করার আগে confusion

#### 2. weighted entropy(children)

👉 split করার পরে average confusion

* ছোট child বেশি influence পাবে
* বড় child বেশি impact রাখে

---

## 🧩 Core intuition:

> Information Gain = Before confusion − After confusion

# 🌳 3. Tree building = Optimization Problem

Decision Tree does:

> Find split S such that impurity is minimized

## 🔵 Impurity measure (Gini)

Binary case:

\text{Gini} = 1 - (p_0^2 + p_1^2)

where:

* (p_1 = P(Play=1))
* (p_0 = P(Play=0))

---

## 🔵 Information Gain form

IG = H(parent) - \sum_{k} \frac{n_k}{n} H(child_k)

# 🧠 1. Fundamental Question: What are we learning?

Machine Learning-এর core problem:

> We are NOT trying to memorize data
> আমরা চাই pattern শিখতে

Mathematically:

[
f(x) \approx y
]

But real goal:

[
\text{minimize generalization error}
]

---

## 🎯 Two types of error:

### 🔵 Training error:

[
E_{train} = \frac{1}{n} \sum (y - \hat{y})^2
]

### 🔴 Test error:

[
E_{test} = \mathbb{E}_{new\ data}[(y - \hat{y})^2]
]

---

# ⚠️ 2. Overfitting — formal definition

Overfitting happens when:

[
E_{train} \downarrow \quad \text{but} \quad E_{test} \uparrow
]

👉 meaning:

* model learns noise + signal
* instead of only signal

---

## 🧠 Scientific intuition:

True function:

[
y = f(x) + \epsilon
]

where:

* (f(x)) = true pattern
* (\epsilon) = noise

👉 Overfitting means:

> model learns (f(x) + \epsilon)

instead of just:

> (f(x))

---

# 🌳 3. Why Decision Trees overfit?

Decision Tree is:

> piecewise constant function approximator

It keeps splitting until purity is achieved:

[
Gini = 0 \quad \text{or} \quad Entropy = 0
]

---

## ⚠️ Problem:

If we allow unlimited depth:

* each leaf becomes 1 sample
* model becomes:

[
f(x_i) = y_i
]

👉 this is **perfect memorization**

---

## 🧠 Geometric interpretation:

Feature space:

* tree creates tiny rectangles
* each rectangle = one label

So:

> training data becomes “lookup table”

---

# 🔴 Post-pruning (Cost Complexity Pruning)

First build full tree (T_0), then prune.

---

## Objective function:

[
R_\alpha(T) = R(T) + \alpha |T|

where:

* (R(T)) = training error
* (|T|) = number of leaves
* (\alpha) = complexity penalty

---

## 🧠 intuition:

> balance accuracy vs simplicity
