```python
import numpy as np

# ডেটাসেট
data = [45, 48, 50, 52, 55, 58, 60, 62, 65, 68, 70, 72, 75, 78, 80, 85, 90, 95, 150, 500]

# রেঞ্জ
range_val = max(data) - min(data)

# গড় (মিন)
mean = np.mean(data)

# পপুলেশন ভ্যারিয়েন্স এবং স্ট্যান্ডার্ড ডেভিয়েশন
var_pop = np.var(data, ddof=0)
std_pop = np.std(data, ddof=0)

# Q2 (মিডিয়ান)
q2 = np.median(data)

# Q1 এবং Q3 (লিনিয়ার ইন্টারপোলেশন মেথড)
q1 = np.percentile(data, 25, method='linear')
q3 = np.percentile(data, 75, method='linear')

# IQR
iqr = q3 - q1

# লোয়ার এবং আপার ফেন্স
lower_fence = q1 - 1.5 * iqr
upper_fence = q3 + 1.5 * iqr

# আউটলায়ার সনাক্তকরণ
outliers = [x for x in data if x < lower_fence or x > upper_fence]

# ফলাফল প্রিন্ট
print(f"রেঞ্জ: {range_val}")
print(f"ভ্যারিয়েন্স (পপুলেশন): {var_pop:.2f}")
print(f"স্ট্যান্ডার্ড ডেভিয়েশন (পপুলেশন): {std_pop:.2f}")
print(f"Q2 (মিডিয়ান): {q2}")
print(f"IQR: {iqr}")
print(f"লোয়ার ফেন্স: {lower_fence}")
print(f"আপার ফেন্স: {upper_fence}")
print(f"আউটলায়ার: {outliers}")
print(f"আউটলায়ারের সংখ্যা: {len(outliers)}")
```
