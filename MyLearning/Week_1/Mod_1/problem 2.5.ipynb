### 🎲 Module 2 – Probability Basics for ML (Practice)

#### Part A – Basic Probability (প্রবলেম ১)

**প্রশ্ন:** ১০০ জনের সার্ভে:
- ৪০ জন pizza পছন্দ করে (A)
- ৫০ জন burger পছন্দ করে (B)
- ২০ জন দুটোই পছন্দ করে (A ∩ B)

খুঁজতে হবে: P(A), P(B), P(A ∩ B), এবং P(A | B)

**ধাপে ধাপে ব্যাখ্যা:**

1. Total people = 100

2. **Formula for simple probability:**
   ```
   P(Event) = (Number of favorable outcomes) / (Total outcomes)
   ```

3. ক্যালকুলেশন:
   - P(A) = 40 / 100 = **0.4** (অর্থাৎ ৪০% লোক pizza পছন্দ করে)
   - P(B) = 50 / 100 = **0.5**
   - P(A ∩ B) = 20 / 100 = **0.2** (দুটোই পছন্দ করে)

4. **Conditional probability formula:**
   ```
   P(A | B) = P(A ∩ B) / P(B)
   ```

5. P(A | B) = 0.2 / 0.5 = **0.4**

**অর্থ কী?** যদি আমরা জানি কেউ burger পছন্দ করে, তাহলে তার pizza পছন্দ করার সম্ভাবনা ৪০%। এটা PDF-এর "Conditional Probability" সেকশনের সাথে মিলে যায়। ML-এ এটা feature দেখে class predict করতে সাহায্য করে।

#### Part B – Bayes’ Theorem (প্রবলেম ২ & ৩)

**প্রবলেম ২:** ১,০০০ ইমেল:
- ১০০ spam (S)
- এর মধ্যে ৪০টিতে “free” শব্দ আছে (F)
- non-spam-এর মধ্যে ৬০টিতেও “free” আছে

খুঁজতে হবে P(S | F)

**Bayes’ Theorem formula (PDF-এ আছে):**
```
P(S | F) = [P(F | S) × P(S)] / P(F)
```

**ধাপে ধাপে:**

1. P(S) = 100 / 1000 = **0.1**
2. P(F | S) = 40 / 100 = **0.4** (spam হলে “free” থাকার সম্ভাবনা)
3. Total “free” containing emails = 40 + 60 = 100 → P(F) = 100 / 1000 = **0.1**
4. P(S | F) = (0.4 × 0.1) / 0.1 = **0.4**

**অর্থ কী?** যদি একটা ইমেলে “free” শব্দ থাকে, তাহলে spam হওয়ার সম্ভাবনা ৪০%। Spam filter-এ এটা important কারণ শুধু “free” দেখে spam বললে অনেক ভুল হবে। PDF-এর base-rate effect-এর মতো—spam কম, তাই false alarm বেশি হতে পারে।

**প্রবলেম ৩:** Disease test
- P(D) = 1/10,000 = **0.0001**
- P(+ | D) = 0.99 (যদি disease থাকে তাহলে test positive)
- P(− | ¬D) = 0.99 → P(+ | ¬D) = 0.01

খুঁজতে হবে P(D | +)

**Bayes’ Theorem again:**
```
P(D | +) = [P(+ | D) × P(D)] / P(+)
```

**ধাপে ধাপে:**

1. P(¬D) = 1 − 0.0001 = 0.9999
2. P(+) = P(+ | D)×P(D) + P(+ | ¬D)×P(¬D)
   = (0.99 × 0.0001) + (0.01 × 0.9999)
   = 0.000099 + 0.009999 ≈ **0.010098**
3. P(D | +) = (0.99 × 0.0001) / 0.010098 ≈ **0.0098** বা প্রায় **0.98%**

**Base-rate effect (PDF-এর exact example):** Disease খুব rare, তাই test ৯৯% accurate হলেও positive result পেলে আসলে disease থাকার সম্ভাবনা মাত্র ১% এর কাছাকাছি। এটা ML-এ class imbalance-এর সাথে সরাসরি যুক্ত।

#### Part C – Confusion Matrix and Metrics

**Task 1:** Confusion Matrix

|                  | Predicted Positive | Predicted Negative |
|------------------|--------------------|--------------------|
| Actual Positive  | 120 (TP)           | 30 (FN)            |
| Actual Negative  | 80 (FP)            | 770 (TN)           |

Total = 1000

**সব formula একসাথে (PDF-এ Sensitivity, Specificity, PPV, NPV আছে):**

```
Accuracy    = (TP + TN) / Total
Precision (PPV) = TP / (TP + FP)
Recall (Sensitivity) = TP / (TP + FN)
F1          = 2 × (Precision × Recall) / (Precision + Recall)
Specificity = TN / (TN + FP)
NPV         = TN / (TN + FN)
Prevalence  = (TP + FN) / Total
```

**ক্যালকুলেশন ধাপে ধাপে:**

1. Accuracy = (120 + 770) / 1000 = **0.89** বা ৮৯%
2. Precision = 120 / (120 + 80) = 120 / 200 = **0.60**
3. Recall = 120 / (120 + 30) = 120 / 150 = **0.80**
4. F1 = 2 × (0.60 × 0.80) / (0.60 + 0.80) = 0.96 / 1.4 ≈ **0.686**
5. Specificity = 770 / (770 + 80) = 770 / 850 ≈ **0.906**
6. NPV = 770 / (770 + 30) = 770 / 800 = **0.9625**
7. Prevalence = 150 / 1000 = **0.15**

**প্রত্যেকটা কী বলে?**
- Accuracy: overall কতটা ঠিক
- Precision: positive বললে কতবার সত্যি positive
- Recall: আসল positive-গুলোর কতটা ধরা পড়ে (important for disease detection)
- F1: precision আর recall-এর balance
- Specificity: negative-গুলো ঠিক ধরা
- NPV: negative বললে কতটা নিশ্চিত negative

**Task 2:** Imbalanced data – 10,000 samples, only 2% positive (200 positive, 9800 negative)
Model সবকিছু negative predict করে।

**Confusion Matrix:**

|                  | Predicted Positive | Predicted Negative |
|------------------|--------------------|--------------------|
| Actual Positive  | 0                  | 200                |
| Actual Negative  | 0                  | 9800               |

**Metrics:**
- Accuracy = 9800 / 10000 = **0.98** (৯৮%)
- Precision = undefined বা 0 (কোনো positive predict করেনি)
- Recall = 0
- F1 = 0
- Specificity = 1

**কেন Accuracy misleading?** Accuracy খুব high কারণ negative class অনেক বড়। কিন্তু model আসলে কিছুই শিখেনি—শুধু majority class predict করে। PDF-এর "Class imbalance" example-এর মতো (fraud detection)।
**কোন metric better?** Recall বা F1 better কারণ minority class (positive) focus করে।
